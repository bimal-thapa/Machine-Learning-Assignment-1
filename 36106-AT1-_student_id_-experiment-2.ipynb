### E.1 Import Algorithm

> Rationale: Importantly, the Ridge Regression algorithm is important in studying regularization techniques that may avoid overfitting and improve generalization of the model. Ridge Regression adds a penalty term toward the complexity of the model, hence often outperforming or providing better stability than ordinary linear regression does. This approach enables the systematic study of various alpha values concerning their impact on model performance.

```
# <fill_this>
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_squared_error
```

<hr>

### E.2 Set Hyperparameters

> Rationale: Import the Ridge Regression algorithm, which will be used to explore some regularization techniques that can be used to avoid overfitting and improve model generalization. By adding a penalty term, Ridge regression controls the model complexity; therefore, it often outperforms and gives more stability than a simple linear regression method. In this way, a systematic study of various alpha values on the performance of the model is possible.

```
# <fill_this>
alpha_values = [0.01, 0.1, 1, 10, 100, 1000]  # Different values of alpha
```

<hr>

### E.3 Fit Model

```
# <fill_this>
# Dictionary to store RMSE results
rmse_results = {}

for alpha in alpha_values:
    # Initialize the Ridge regression model with a specific alpha value
    ridge_model = Ridge(alpha=alpha)
    
    # Fit the model to the training data
    ridge_model.fit(X_train[features_list], y_train)
    
    # Predict on validation data
    y_pred_val = ridge_model.predict(X_val[features_list])
    
    # Calculate RMSE on validation set
    rmse_val = mean_squared_error(y_val, y_pred_val, squared=False)
    
    # Store the RMSE for the current alpha
    rmse_results[alpha] = rmse_val

# Print the RMSE for each alpha value
for alpha, rmse_val in rmse_results.items():
    print(f"Alpha: {alpha}, RMSE on validation set: {rmse_val}")
```

<hr>

### E.4 Model Technical Performance

```
# <fill_this>
# Find the best alpha value with the lowest RMSE
best_alpha = min(rmse_results, key=rmse_results.get)
best_rmse = rmse_results[best_alpha]

print(f"Best alpha: {best_alpha} with RMSE: {best_rmse}")
```

> Results: The best alpha value for Ridge Regression was found to be 100, which resulted in an RMSE of 4993.69 on the validation set. This indicates that with this level of regularization, the model achieved a balance between fit and complexity, leading to a reasonably accurate performance.

### E.5 Business Impact from Current Model Performance

```
# Ensure mean_salary is a scalar (convert to float if necessary)
mean_salary = float(y_train.mean())  # Convert to float to ensure it's a scalar value

# Ensure best_rmse is a scalar (convert to float if necessary)
best_rmse = float(best_rmse)  # Convert to float to ensure it's a scalar value

# Business Impact:
print(f"Mean salary in the training set: {mean_salary}")

# Compare the RMSE for the best alpha to the mean salary
print(f"Best model RMSE accounts for {best_rmse / mean_salary:.2%} of the average salary")

# Recommendations:
if best_rmse > mean_salary * 0.1:  # If error is more than 10% of average salary
    print("Recommendation: The error is too high. Consider further feature engineering or different models.")
else:
    print("Recommendation: The model is performing well; slight refinements may improve business impact.")
```
