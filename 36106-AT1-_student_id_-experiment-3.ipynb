### C.1   Load Datasets

## E. Train Machine Learning Model

```
train_model_executive_summary = 'Therefore, we conducted multiple trainings of the Lasso Linear Regression model with various values of alpha to see the impact this hyperparameter has on the performance of our model. All these results are in the quest for getting the best alpha, hence the model complexity is well balanced against its predictive accuracy.'
```

### E.1 Import Algorithm

> Rationale: In this work, we adopt the Lasso algorithm because it can carry out variable selection and regularization to prevent overfitting, which will simultaneously improve model interpretability. We import this algorithm to evaluate how different alpha values can affect the performance of models and feature importance.

```
# <fill_this>
from sklearn.linear_model import Lasso
from sklearn.metrics import mean_squared_error
```

<hr>

### E.2 Set Hyperparameters

> Rationale: Setting different values of alpha for the Lasso model will balance fitting the training data well and model simplicity, in a nutshell. By having these, one can explore better performance for different values of alpha, looking for an optimum alpha that minimizes overfitting.

```
# <fill_this>
alphas = [0.01, 0.1, 1, 10, 100, 1000]  # Example values for alpha
```

<hr>

### E.3 Fit Model

```
# Initialize dictionaries to store results
rmse_results = {}

# Fit models with different alpha values
for alpha in alphas:
    # Initialize the Lasso model with the current alpha value
    model = Lasso(alpha=alpha)
    
    # Fit the model
    model.fit(X_train[features_list], y_train)
    
    # Predict on validation data
    y_pred = model.predict(X_val[features_list])
    
    # Calculate RMSE
    rmse = mean_squared_error(y_val, y_pred, squared=False)
    
    # Store the RMSE result
    rmse_results[alpha] = rmse
    
    print(f"Alpha: {alpha}, RMSE: {rmse}")

# Determine the best alpha based on RMSE
best_alpha = min(rmse_results, key=rmse_results.get)
best_rmse = rmse_results[best_alpha]

print(f"Best alpha: {best_alpha}")
print(f"Best RMSE: {best_rmse}")

# Fit the model with the best alpha on the training set
best_model = Lasso(alpha=best_alpha)
best_model.fit(X_train[features_list], y_train)
```

<hr>

### E.4 Model Technical Performance

```
# <fill_this>
# Predict on the test set using the best model
y_test_pred = best_model.predict(X_test[features_list])

# Calculate RMSE on the test set
test_rmse = mean_squared_error(y_test, y_test_pred, squared=False)
print(f"Test RMSE with best alpha ({best_alpha}): {test_rmse}")


```

> Results: The best-performing Lasso model, with an alpha value of 10, achieved a Test RMSE of 15840.60. This indicates that, on average, the model's salary predictions deviate from the actual values by approximately 15840 units, providing a benchmark for evaluating potential improvements in future experiments.

<hr>

### E.5 Business Impact from Current Model Performance

```
# Ensure that mean_salary, best_rmse, and test_rmse are scalar values
mean_salary = float(y_train.mean())  # Convert to scalar if necessary
best_rmse = float(best_rmse)  # Convert to scalar if necessary
test_rmse = float(test_rmse)  # Convert to scalar if necessary

# Business Impact
print(f"Mean salary in the training set: {mean_salary:.2f}")

# Compare the RMSE for the best alpha to the mean salary
print(f"Best model RMSE accounts for {best_rmse/mean_salary:.2%} of the average salary")
print(f"Test RMSE with best alpha ({best_alpha}) accounts for {test_rmse/mean_salary:.2%} of the average salary")

# Recommendations
if test_rmse > mean_salary * 0.1:  # If error is more than 10% of average salary
    print("Recommendation: The error is too high. Consider further feature engineering or trying different models.")
else:
    print("Recommendation: The model is performing well; slight refinements may improve business impact.")
```
